```python
# Assigns values to variables
secret_number = 17
max_attempts = 5
attempt_number = 0

# Instructs the user what to do
print("There is a secret number! You are given a maximum of 5 valid guesses.")
print("Negativity we ignore. Positivity we prefer. Enter a guess and you may win the game.")

# Leaves spacing for the output to look cleaner
print("")

# Prompts the user to guess the secret number
while attempt_number < max_attempts:
  guess = int(input(f"Attempt {attempt_number + 1} - Guess the number: "))


  # If the user enters a negative number the attempt will be ignored
  # If not the attempt will be recorded
  # It also gives out hints to the user whether the guess is too high or too low
  if 0 < guess < secret_number:
    print("Too low!")
    attempt_number += 1
  elif guess == 17:
    print("Congratulations! You have guessed the right number!")
    break
  elif guess < 0:
    print("Negative number ignored")
  elif guess > secret_number:
    print("Too High!")
    attempt_number += 1

# Displays a message if the secret number is not guessed correctly
if attempt_number == max_attempts:
  print("")
  print("You have run out of valid attempts. Now you'll NEVER know the secret number.")




```
